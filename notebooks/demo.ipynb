# RAG Pipeline Demo - Financial Document Q&A

This notebook demonstrates the RAG (Retrieval-Augmented Generation) pipeline for querying financial documents.

**What you'll learn:**
- How to ingest and index documents
- How embeddings and vector search work
- How to query the system
- How to inspect retrieved context

## Setup

```python
import sys
from pathlib import Path
import logging

# Add src to path
sys.path.insert(0, str(Path().absolute().parent))

# Configure logging
logging.basicConfig(level=logging.INFO)

from src.rag_pipeline import RAGPipeline
from dotenv import load_dotenv

# Load environment variables (for LLM API keys)
load_dotenv()
```

## Initialize Pipeline

```python
# Initialize RAG pipeline
rag = RAGPipeline(
    data_dir=Path("../data"),
    top_k=3  # Retrieve top 3 most relevant chunks
)

print("✅ Pipeline initialized")
```

## Ingest Documents

This step:
1. Loads documents from `data/` directory
2. Splits them into chunks
3. Generates embeddings using sentence-transformers
4. Stores embeddings in FAISS vector store

```python
# Ingest documents (force=True to re-index)
num_chunks = rag.ingest_documents(force=True)
print(f"\n✅ Indexed {num_chunks} chunks")
```

## Example Queries

```python
# Query 1: Capital requirements
question1 = "What are the minimum capital requirements under Basel III?"
answer1 = rag.query(question1)

print("\n" + "="*80)
print(f"QUESTION: {question1}")
print("="*80)
print(answer1)
```

```python
# Query 2: Liquidity ratios
question2 = "Explain the Liquidity Coverage Ratio (LCR)"
answer2 = rag.query(question2)

print("\n" + "="*80)
print(f"QUESTION: {question2}")
print("="*80)
print(answer2)
```

```python
# Query 3: Leverage ratio
question3 = "What is the minimum leverage ratio and why is it important?"
answer3 = rag.query(question3)

print("\n" + "="*80)
print(f"QUESTION: {question3}")
print("="*80)
print(answer3)
```

## Inspect Retrieved Chunks

Let's see what chunks were retrieved for a query:

```python
# Retrieve chunks without generating answer
test_query = "What are systemically important banks?"
chunks = rag.retriever.retrieve(test_query)

print(f"Retrieved {len(chunks)} chunks for: '{test_query}'\n")

for i, chunk in enumerate(chunks, 1):
    print(f"\n{'='*80}")
    print(f"CHUNK {i}")
    print(f"{'='*80}")
    print(f"Source: {chunk['metadata']['filename']}")
    print(f"Chunk ID: {chunk['metadata']['chunk_id']}")
    print(f"\nContent (first 300 chars):\n{chunk['content'][:300]}...")
```
